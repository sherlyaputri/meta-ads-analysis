```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from causalimpact import CausalImpact

# ==========================================
# 1. BACA DATA DARI FOLDER LOKAL WAHANA
# ==========================================
# Ganti dengan path folder wahana di laptopmu
FILE_PATH = r'C:\Users\Wahana Express\Documents\wahana\ads\master_ads.xlsx' 
# (Atau sesuaikan nama file Excel-mu, misal: 'laporan_ads_oktober.xlsx')

print("Membaca data dari Excel...")
df = pd.read_excel(FILE_PATH)

# ==========================================
# 2. DATA CLEANING (SANGAT PENTING)
# ==========================================
# A. Membersihkan kolom Week: Ambil tanggal awal dari format "04/10/25 - 10/10/25"
# Pisahkan berdasarkan tanda strip "-", ambil bagian kiri [0], lalu hapus spasi
df['Tanggal_Mulai'] = df['Week'].str.split('-').str[0].str.strip()

# Ubah string tanggal menjadi format Datetime (Asumsi format DD/MM/YY dari datamu)
df['Tanggal_Mulai'] = pd.to_datetime(df['Tanggal_Mulai'], format='%d/%m/%y')

# Jadikan 'Tanggal_Mulai' sebagai Index dataframe
df.set_index('Tanggal_Mulai', inplace=True)

# Urutkan data dari tanggal terlama ke terbaru (wajib untuk Time Series)
df = df.sort_index()

# B. Membersihkan kolom Actual Spend agar jadi angka murni (jika terbaca sebagai string/teks)
# Hapus "Rp", spasi, dan koma. Jika dari awal pandas membacanya sebagai angka, ini tetap aman.
df['Actual Spend'] = df['Actual Spend'].astype(str)\
                       .str.replace('Rp', '', regex=False)\
                       .str.replace(',', '', regex=False)\
                       .str.replace(' ', '', regex=False)\
                       .astype(float)

# C. Pastikan kolom Jumlah TTK adalah angka numerik
df['Jumlah TTK'] = pd.to_numeric(df['Jumlah TTK'], errors='coerce')

# ==========================================
# 3. PERSIAPAN MODEL CAUSAL IMPACT
# ==========================================
# Kita hanya butuh target kolom yang mau dianalisis: 'Jumlah TTK'
data_model = df[['Jumlah TTK']].copy()

# TENTUKAN PERIODE (Ganti tanggal ini sesuai datamu)
# Format tanggal saat dipanggil harus YYYY-MM-DD
# Misalnya, minggu ke-1 & ke-2 iklan mati (Baseline), minggu ke-3 & ke-4 iklan nyala
pre_period = ['2025-10-04', '2025-10-11']   # Periode Iklan MATI
post_period = ['2025-10-20', '2025-10-27']  # Periode Iklan NYALA

print(f"Menjalankan model Causal Impact...")
# Inisialisasi dan jalankan model
impact = CausalImpact(data_model, pre_period, post_period)

# ==========================================
# 4. HASIL ANALISIS (WORTH IT ATAU ENGGA)
# ==========================================
# Ambil angka "Uplift" (Tambahan TTK murni karena iklan)
total_uplift_ttk = impact.summary_data.loc['abs_effect', 'cumulative']

# Hitung total spend di periode saat iklan nyala
total_spend = df.loc[post_period[0]:post_period[1], 'Actual Spend'].sum()

# Masukkan margin keuntungan per resi/TTK (misal Rp 5.000)
MARGIN_PER_TTK = 5000 
estimasi_pendapatan_iklan = total_uplift_ttk * MARGIN_PER_TTK
net_profit_iklan = estimasi_pendapatan_iklan - total_spend

print("\n" + "="*50)
print("HASIL KESIMPULAN")
print("="*50)
print(f"Total Biaya Iklan (Actual Spend)     : Rp {total_spend:,.0f}")
print(f"Estimasi Tambahan TTK Murni (Uplift) : {int(total_uplift_ttk):,} resi")
print(f"Estimasi Profit Kotor dari Iklan     : Rp {estimasi_pendapatan_iklan:,.0f}")
print(f"Net Profit Iklan (Setelah dikurangi) : Rp {net_profit_iklan:,.0f}")

if net_profit_iklan > 0:
    print("STATUS: IKLAN INI WORTH IT! (Mendatangkan profit)")
else:
    print("STATUS: IKLAN TIDAK WORTH IT! (Lebih besar pasak daripada tiang)")
print("="*50)

# Tampilkan visualisasi grafik langsung di JupyterLab
impact.plot()
plt.show()

# Tampilkan laporan statistik detail
print(impact.summary())
```
